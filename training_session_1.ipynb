#### Defining arrays

Array handling is conveniently done in the package [NumPy](https://numpy.org/).
To use this package it has to be imported before using it in the code using:

```python
import numpy as np
```

Define an array by:

```python
a = np.array([[1, 1, 2], [3, 4, 5]])  # row 1  # row 2
# printing variable can be done using a formatting string like this.
# variable are placed in curly brackets
print(f"a: {a}")
```

An array of shape *2 x 3* with zeros with zeros can be defined like this:

```python
b = np.zeros(shape=(2, 3))
```

Set the array value of the first column and first row to *5* (Indexing in Python starts from *0*):

```python
b[0, 0] = 5
print(f"b: {b}")
```

Get values of second column from *a* and set it to last column of *b*:

```python
b[:, 2] = a[:, 1]
print(f"b: {b}")
```

Get dimensions of array *b* and print result with:

```python
print(f"shape of b: {b.shape}")
```

Component wise summation (+), subtraction (-), multiplication (*), division (/), potentiation (a**b) of arrays with the same dimensions:

```python
sum_a_b = a + b
diff_a_b = a - b
product_a_b = a * b
frac_b_a = b / a
potentiation_a_b = a**b
```

Scalar product of 𝐴 ∈ ℝ$^{𝑛×𝑚}$ and B ∈ ℝ$^{i×j}$:

```python
a = np.array([[1, 2], [2, 3], [1, 2]])
b = np.ones(shape=(2, 3))
dot_product_a_b = a.dot(b)
```

Cross product of 𝐴 ∈ ℝ$^𝑛$ and B ∈ ℝ$^i$:

```python
a = np.array([1, 2, 3])
b = np.ones(3)
cross_product_a_b = np.cross(a, b)
```

```python
i = 0
while i <= 3:
    print(i)
    i = i + 1  # same as i += 1
```

#### Solving a system of algebraic equations

Using [fsolve](https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.fsolve.html) method from [scipy.optimize](https://docs.scipy.org/doc/scipy/reference/optimize.html) to solve a system of (non-linear) algebraic 
equations

```python
from scipy.optimize import fsolve  # type: ignore
```

Defining the AEs system:
- AEs system needs to be defined as function. 
- Searched variables are the input of the function
- Equations, rearranged as root-problem are the output of the function

Defining the AE functions:

```python
def AEs(x):
    eq0 = 3 * x[0] - x[1] + 1
    eq1 = x[0] + x[0] * x[1] - 2
    return eq0, eq1
```

Solving our defined AE systems with an initial guess *x_0*:

```python
x = fsolve(func=AEs, x0=[0, 0])
print(x)
print(AEs(x))
```

#### Plotting data

A common library for visualizing data is [Matplotlib](https://matplotlib.org/) with the module [pyplot](https://matplotlib.org/3.5.3/api/_as_gen/matplotlib.pyplot.html)

##### General workflow to generate a plot

Prepare data for plotting. Usually vector *y* is plotted against *x*. Both have the same length.
Here some test data is generated using numpy arrays:

```python
import numpy as np

k = 1  # [1/s]
c = np.linspace(0, 10, 11)  # [mol/m^3]
n = 2
r = k * c**n
```

Then [pyplot](https://matplotlib.org/3.5.3/api/_as_gen/matplotlib.pyplot.html) is imported:

```python
import matplotlib.pyplot as plt
```

```python
# Create figure and axes of plot
fig, ax = plt.subplots(1, 1, figsize=(4, 3))

# Plot values
ax.plot(c, r, color="green", linestyle="solid", marker="o", label="$n = 2$")

# Set axes parameters
ax.set_xlabel("$c_i \\; \\mathregular{/mol \\, m^{-3}}$")
ax.set_ylabel("$r \\; \\mathregular{/mol \\, s^{-1} m{-3}}$")
ax.set_xlim(0, 10)
ax.legend(fancybox=False, loc="best", edgecolor="black")
```

```python
# Save figure to a file
filepath = "figure.svg"
fig.savefig(filepath, dpi=600, pad_inches=0.1, bbox_inches="tight")
```

Generating a plot with three subplots with different reaction orders *n* can look like this:

```python
# Create figure and array of axes of plot
fig, axs = plt.subplots(3, 1, figsize=(4, 8), sharex=True)
ns = np.array([0.5, 1, 2])
colors = ["red", "green", "blue"]
# "enumerate" can be used to loop over entries of an iterable and keep count of the index
for i, n in enumerate(ns):
    r = k * c**n
    axs[i].plot(
        c, r, color=colors[i], linestyle="solid", marker="o", label=f"$n = {n}$"
    )
    axs[i].set_ylabel("$r \\; \\mathregular{/mol \\, s^{-1} m^{-3}}$")
    axs[i].legend(fancybox=False, loc="best", edgecolor="black")
axs[2].set_xlabel("$c_{i} \\; \\mathregular{/mol \\, m^{-3}}$")
```
